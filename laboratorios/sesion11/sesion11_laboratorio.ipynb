### 📄 `app.py`

### 📄 `repositorio.py`

Capa de acceso a datos.

````python
%%writefile repositorio.py
"""Capa de DATOS: guarda y recupera libros en memoria. No conoce HTTP ni FastAPI."""
from dataclasses import asdict, dataclass


@dataclass
class LibroRegistro:
    id: int
    titulo: str
    anio: int


class RepositorioLibros:
    def __init__(self) -> None:
        self._libros: dict[int, LibroRegistro] = {}
        self._siguiente_id = 1

    def agregar(self, titulo: str, anio: int) -> dict:
        registro = LibroRegistro(id=self._siguiente_id, titulo=titulo, anio=anio)
        self._libros[registro.id] = registro
        self._siguiente_id += 1
        return asdict(registro)

    def listar(self) -> list[dict]:
        return [asdict(r) for r in self._libros.values()]

    def existe_titulo(self, titulo: str) -> bool:
        return any(r.titulo.casefold() == titulo.casefold() for r in self._libros.values())
````

### 📄 `servicio.py`

Capa de lógica de negocio.

````python
%%writefile servicio.py
"""Capa de NEGOCIO: reglas para crear y listar libros. No conoce HTTP."""
from repositorio import RepositorioLibros


class TituloDuplicado(Exception):
    """Regla de negocio: no puede haber dos libros con el mismo título."""


class ServicioLibros:
    def __init__(self, repositorio: RepositorioLibros) -> None:
        self._repo = repositorio

    def crear(self, titulo: str, anio: int) -> dict:
        titulo = " ".join(titulo.split())          # normaliza espacios internos
        if self._repo.existe_titulo(titulo):
            raise TituloDuplicado(titulo)
        return self._repo.agregar(titulo, anio)

    def listar(self) -> list[dict]:
        return sorted(self._repo.listar(), key=lambda libro: libro["id"])
````

### 📄 `router.py`

Capa de exposición HTTP.

````python
%%writefile router.py
"""Capa HTTP: valida el DTO, delega al servicio y traduce errores a códigos HTTP."""
from fastapi import APIRouter, HTTPException, status
from pydantic import BaseModel, ConfigDict, Field

from repositorio import RepositorioLibros
from servicio import ServicioLibros, TituloDuplicado

router = APIRouter(prefix="/api/libros", tags=["libros"])

# Ensamblado de las capas (en la sesión 12 esto se hará con Depends)
_servicio = ServicioLibros(RepositorioLibros())


class LibroEntrada(BaseModel):
    model_config = ConfigDict(extra="forbid")
    titulo: str = Field(..., min_length=1, max_length=200)
    anio: int = Field(..., ge=1450, le=2100)


class Libro(LibroEntrada):
    id: int


class ListaLibros(BaseModel):
    libros: list[Libro]


@router.get("", response_model=ListaLibros)
def listar() -> ListaLibros:
    return ListaLibros(libros=_servicio.listar())


@router.post("", response_model=Libro, status_code=status.HTTP_201_CREATED)
def crear(datos: LibroEntrada) -> dict:
    try:
        return _servicio.crear(datos.titulo, datos.anio)
    except TituloDuplicado as exc:
        raise HTTPException(status.HTTP_409_CONFLICT, f"Ya existe el libro «{exc}»") from exc
````
