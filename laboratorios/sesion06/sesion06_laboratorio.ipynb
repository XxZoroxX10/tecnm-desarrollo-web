### 📄 `app.py`

Archivo principal de la implementación.

````python
%%writefile app.py
"""Sesión 06 · Reglas de dominio en el DTO y paginación validada en la query."""
from typing import Annotated

from fastapi import FastAPI, Query, status
from pydantic import BaseModel, ConfigDict, Field, StringConstraints

app = FastAPI(title="Límites de negocio · DAB-2201", version="1.0.0")

# strip_whitespace: "   Rayuela  " -> "Rayuela"; así "   " cuenta como vacío (< 3)
Titulo = Annotated[str, StringConstraints(strip_whitespace=True, min_length=3, max_length=80)]


class LibroEntrada(BaseModel):
    model_config = ConfigDict(extra="forbid")
    titulo: Titulo
    anio: int = Field(..., ge=1450, le=2100)     # 1450: aprox. la imprenta de Gutenberg
    precio: float = Field(..., gt=0, le=100_000)  # gt=0: el 0 NO es válido


class Libro(LibroEntrada):
    id: int


_libros: list[Libro] = []


@app.post("/api/libros", status_code=status.HTTP_201_CREATED, response_model=Libro)
def crear(datos: LibroEntrada) -> Libro:
    libro = Libro(id=len(_libros) + 1, **datos.model_dump())
    _libros.append(libro)
    return libro


@app.get("/api/libros")
def listar(
    limite: int = Query(10, ge=1, le=100, description="Cuántos libros devolver (1-100)"),
    desde: int = Query(0, ge=0, description="Cuántos libros saltar"),
) -> dict:
    pagina = _libros[desde: desde + limite]
    return {"libros": pagina, "total": len(_libros), "limite": limite, "desde": desde}
````
