### 📄 `app.py`

Archivo principal de la implementación.

````python
%%writefile app.py
"""Sesión 08 · Errores semánticos con un sobre único: {"error": {"codigo", "mensaje"}}."""
import logging

from fastapi import FastAPI, Request, status
from fastapi.exceptions import RequestValidationError
from fastapi.responses import JSONResponse
from pydantic import BaseModel, Field
from starlette.exceptions import HTTPException as StarletteHTTPException

app = FastAPI(title="Errores semánticos · DAB-2201", version="1.0.0")
log = logging.getLogger("biblioteca")


# ---------- Excepción de negocio ----------
class LibroNoEncontrado(Exception):
    def __init__(self, libro_id: int):
        self.libro_id = libro_id


def sobre(codigo: str, mensaje: str, detalles: list | None = None) -> dict:
    error = {"codigo": codigo, "mensaje": mensaje}
    if detalles:
        error["detalles"] = detalles
    return {"error": error}


# ---------- Manejadores globales ----------
@app.exception_handler(LibroNoEncontrado)
async def libro_no_encontrado(request: Request, exc: LibroNoEncontrado):
    return JSONResponse(status_code=404,
                        content=sobre("NO_ENCONTRADO", f"El libro {exc.libro_id} no existe"))


@app.exception_handler(RequestValidationError)
async def error_validacion(request: Request, exc: RequestValidationError):
    # Solo campo + mensaje: no devolvemos el input completo para no reflejar datos del cliente
    detalles = [{"campo": ".".join(str(p) for p in e["loc"][1:]), "mensaje": e["msg"]}
                for e in exc.errors()]
    return JSONResponse(status_code=422,
                        content=sobre("VALIDACION", "Los datos enviados no son válidos", detalles))


@app.exception_handler(StarletteHTTPException)
async def error_http(request: Request, exc: StarletteHTTPException):
    # 404 de ruta inexistente, 405, etc. con el mismo sobre
    codigos = {404: "NO_ENCONTRADO", 405: "METODO_NO_PERMITIDO"}
    return JSONResponse(status_code=exc.status_code, headers=getattr(exc, "headers", None),
                        content=sobre(codigos.get(exc.status_code, "ERROR_HTTP"), str(exc.detail)))


@app.exception_handler(Exception)
async def error_inesperado(request: Request, exc: Exception):
    # Último recurso: se registra en el log, pero al cliente NO le llega el traceback
    log.exception("Error no controlado")
    return JSONResponse(status_code=500, content=sobre("ERROR_INTERNO", "Ocurrió un error inesperado"))


# ---------- Modelos y datos ----------
class LibroEntrada(BaseModel):
    titulo: str = Field(..., min_length=1, max_length=200)
    anio: int = Field(..., ge=1450, le=2100)


class Libro(LibroEntrada):
    id: int


_libros: dict[int, Libro] = {1: Libro(id=1, titulo="Rayuela", anio=1963)}


@app.get("/api/libros/{libro_id}", response_model=Libro)
def obtener(libro_id: int) -> Libro:
    if libro_id not in _libros:
        raise LibroNoEncontrado(libro_id)
    return _libros[libro_id]


@app.post("/api/libros", status_code=status.HTTP_201_CREATED, response_model=Libro)
def crear(datos: LibroEntrada) -> Libro:
    nuevo = Libro(id=max(_libros, default=0) + 1, **datos.model_dump())
    _libros[nuevo.id] = nuevo
    return nuevo
````
