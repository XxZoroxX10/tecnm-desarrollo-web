**Requisitos del contrato**

## 🧩 El reto: escribe el resultado

Crea el o los archivos del contrato. No hay una plantilla de pasos: el esqueleto solo te da el punto de partida. Todo lo que sigue lo decides tú.

**Archivos que debes producir:** `app.py`


### 📄 `app.py`

Archivo principal de la implementación.

````python
%%writefile app.py
"""Sesión 01 · Diagnóstico HTTP: un servidor que responde JSON y no se cae."""
from fastapi import FastAPI
from pydantic import BaseModel, Field

app = FastAPI(title="Diagnóstico HTTP · DAB-2201", version="1.0.0")


class Eco(BaseModel):
    """Lo que el cliente nos manda en el body (JSON)."""
    mensaje: str = Field(..., min_length=1, max_length=500)


class EcoRespuesta(BaseModel):
    mensaje: str
    recibido: bool


@app.get("/api/estado")
def estado() -> dict:
    # Endpoint de salud: si responde, el servidor está vivo.
    return {"servidor": "ok", "curso": "DAB-2201"}


@app.post("/api/eco", response_model=EcoRespuesta)
def eco(datos: Eco) -> EcoRespuesta:
    # Pydantic ya validó el JSON: si viene roto o sin "mensaje", FastAPI responde 422
    # automáticamente y el servidor sigue de pie.
    return EcoRespuesta(mensaje=datos.mensaje, recibido=True)

# GET /api/eco no existe como ruta GET -> FastAPI responde 405 Method Not Allowed
# (la ruta sí existe, el verbo no), con el encabezado Allow: POST.
````
