**Requisitos del contrato**

- Un origen no autorizado no recibe access-control-allow-origin.

> **Pista conceptual:** `CORSMiddleware` con `allow_origins=[...]` es una lista blanca. Un `*` en producción equivale a invitar a cualquiera.
