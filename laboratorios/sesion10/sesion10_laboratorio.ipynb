### 📄 `app.py`

Archivo principal de la implementación.

````python
%%writefile app.py
"""Sesión 10 · Hito 1: Mock API de libros contract-first (CRUD + validación + OpenAPI + errores)."""
import time
import uuid
from typing import Annotated

from fastapi import FastAPI, Request, Response, status
from fastapi.exceptions import RequestValidationError
from fastapi.responses import JSONResponse
from pydantic import BaseModel, ConfigDict, Field, StringConstraints
from starlette.exceptions import HTTPException as StarletteHTTPException

app = FastAPI(
    title="Biblioteca Mock API",
    version="0.1.0",
    description="Hito 1 · Mock API RESTful con contrato OpenAPI y validación en el borde.",
)


# ---------------- DTOs (contrato) ----------------
Titulo = Annotated[str, StringConstraints(strip_whitespace=True, min_length=1, max_length=120)]


class LibroEntrada(BaseModel):
    model_config = ConfigDict(extra="forbid")
    titulo: Titulo
    anio: int = Field(..., ge=1450, le=2100)


class Libro(LibroEntrada):
    id: int = Field(..., description="Asignado por el servidor")


class ListaLibros(BaseModel):
    libros: list[Libro]


class ErrorCuerpo(BaseModel):
    codigo: str
    mensaje: str


class Error(BaseModel):
    error: ErrorCuerpo


# ---------------- Errores uniformes ----------------
class LibroNoEncontrado(Exception):
    def __init__(self, libro_id: int):
        self.libro_id = libro_id


def sobre(status_code: int, codigo: str, mensaje: str, headers=None) -> JSONResponse:
    return JSONResponse(status_code=status_code, headers=headers,
                        content={"error": {"codigo": codigo, "mensaje": mensaje}})


@app.exception_handler(LibroNoEncontrado)
async def _no_encontrado(_: Request, exc: LibroNoEncontrado):
    return sobre(404, "NO_ENCONTRADO", f"El libro {exc.libro_id} no existe")


@app.exception_handler(RequestValidationError)
async def _validacion(_: Request, exc: RequestValidationError):
    campos = ", ".join(".".join(map(str, e["loc"][1:])) or "body" for e in exc.errors())
    return sobre(422, "VALIDACION", f"Datos inválidos en: {campos}")


@app.exception_handler(StarletteHTTPException)
async def _http(_: Request, exc: StarletteHTTPException):
    codigos = {404: "NO_ENCONTRADO", 405: "METODO_NO_PERMITIDO"}
    return sobre(exc.status_code, codigos.get(exc.status_code, "ERROR_HTTP"), str(exc.detail),
                 headers=getattr(exc, "headers", None))


@app.exception_handler(Exception)
async def _interno(_: Request, exc: Exception):
    return sobre(500, "ERROR_INTERNO", "Ocurrió un error inesperado")   # sin traceback


# ---------------- Middleware de trazabilidad ----------------
@app.middleware("http")
async def trazabilidad(request: Request, call_next):
    inicio = time.perf_counter()
    respuesta = await call_next(request)
    respuesta.headers["X-Request-ID"] = request.headers.get("x-request-id") or str(uuid.uuid4())
    respuesta.headers["X-Process-Time"] = f"{time.perf_counter() - inicio:.6f}"
    return respuesta


# ---------------- Repositorio en memoria ----------------
_libros: dict[int, Libro] = {}
_siguiente_id = 1


def _buscar(libro_id: int) -> Libro:
    if libro_id not in _libros:
        raise LibroNoEncontrado(libro_id)
    return _libros[libro_id]


ERRORES = {404: {"model": Error, "description": "Libro inexistente"},
           422: {"model": Error, "description": "Datos inválidos"}}


# ---------------- Rutas (sustantivos, el verbo HTTP es la acción) ----------------
@app.get("/api/libros", response_model=ListaLibros, tags=["libros"], summary="Listar libros")
def listar() -> ListaLibros:
    return ListaLibros(libros=list(_libros.values()))


@app.post("/api/libros", response_model=Libro, status_code=status.HTTP_201_CREATED,
          tags=["libros"], summary="Crear libro", responses={422: ERRORES[422]})
def crear(datos: LibroEntrada, response: Response) -> Libro:
    global _siguiente_id
    libro = Libro(id=_siguiente_id, **datos.model_dump())
    _libros[libro.id] = libro
    _siguiente_id += 1
    response.headers["Location"] = f"/api/libros/{libro.id}"
    return libro


@app.get("/api/libros/{libro_id}", response_model=Libro, tags=["libros"],
         summary="Obtener libro", responses={404: ERRORES[404]})
def obtener(libro_id: int) -> Libro:
    return _buscar(libro_id)


@app.put("/api/libros/{libro_id}", response_model=Libro, tags=["libros"],
         summary="Reemplazar libro", responses=ERRORES)
def reemplazar(libro_id: int, datos: LibroEntrada) -> Libro:
    _buscar(libro_id)
    _libros[libro_id] = Libro(id=libro_id, **datos.model_dump())
    return _libros[libro_id]


@app.delete("/api/libros/{libro_id}", status_code=status.HTTP_204_NO_CONTENT, tags=["libros"],
            summary="Eliminar libro", responses={404: ERRORES[404]})
def eliminar(libro_id: int) -> Response:
    _buscar(libro_id)
    del _libros[libro_id]
    return Response(status_code=status.HTTP_204_NO_CONTENT)
````
