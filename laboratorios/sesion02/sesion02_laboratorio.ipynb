**Requisitos del contrato**

- GET /api/libros responde 200 con {"libros": [...]}.

> **Pista conceptual:** En FastAPI, si defines GET/PUT/DELETE para `/api/libros/{id}` pero no PATCH, el framework responde 405 automáticamente. No programes ese caso: deja que el contrato hable.

### 📄 `app.py`

Archivo principal de la implementación.

````python
%%writefile app.py
"""Sesión 02 · CRUD de libros en memoria: cada verbo HTTP con su código correcto."""
from fastapi import FastAPI, HTTPException, Response, status
from pydantic import BaseModel, Field

app = FastAPI(title="CRUD de libros · DAB-2201", version="1.0.0")


class LibroEntrada(BaseModel):
    titulo: str = Field(..., min_length=1, max_length=200)
    anio: int = Field(..., ge=0, le=2100)


class Libro(LibroEntrada):
    id: int


# "Base de datos" en memoria: dict id -> Libro, y un contador que nunca reutiliza ids
libros: dict[int, Libro] = {}
siguiente_id = 1


def buscar(libro_id: int) -> Libro:
    if libro_id not in libros:
        raise HTTPException(status_code=404, detail=f"El libro {libro_id} no existe")
    return libros[libro_id]


@app.get("/api/libros")
def listar() -> dict:
    return {"libros": list(libros.values())}


@app.post("/api/libros", status_code=status.HTTP_201_CREATED, response_model=Libro)
def crear(datos: LibroEntrada) -> Libro:
    global siguiente_id
    libro = Libro(id=siguiente_id, **datos.model_dump())   # el id lo asigna el servidor
    libros[libro.id] = libro
    siguiente_id += 1
    return libro


@app.get("/api/libros/{libro_id}", response_model=Libro)
def obtener(libro_id: int) -> Libro:
    return buscar(libro_id)


@app.put("/api/libros/{libro_id}", response_model=Libro)
def reemplazar(libro_id: int, datos: LibroEntrada) -> Libro:
    buscar(libro_id)                                   # 404 si no existe
    libros[libro_id] = Libro(id=libro_id, **datos.model_dump())   # PUT = reemplazo total
    return libros[libro_id]


@app.delete("/api/libros/{libro_id}", status_code=status.HTTP_204_NO_CONTENT)
def eliminar(libro_id: int) -> Response:
    buscar(libro_id)
    del libros[libro_id]
    return Response(status_code=status.HTTP_204_NO_CONTENT)   # 204 = sin cuerpo

# PATCH no está definido: como la ruta /api/libros/{id} sí existe con otros verbos,
# FastAPI responde 405 Method Not Allowed (con encabezado Allow), no 404.
````
