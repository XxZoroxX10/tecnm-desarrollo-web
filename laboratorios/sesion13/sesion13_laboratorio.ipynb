### 📄 `app.py`

Archivo principal de la implementación.

````python
%%writefile app.py
"""Sesión 13 · Sesión con estado: el servidor guarda la sesión; el cliente solo un id opaco en cookie."""
import hmac
import secrets
import time

from fastapi import Cookie, FastAPI, HTTPException, Response, status
from pydantic import BaseModel, Field

app = FastAPI(title="Sesiones stateful · DAB-2201", version="1.0.0")

COOKIE = "session_id"
DURACION_SEG = 30 * 60                       # la sesión expira a los 30 minutos

# Usuario de prueba del laboratorio (en la sesión 14 la contraseña se guardará con hash)
USUARIOS = {"ana": {"password": "secreta", "nombre": "Ana"}}

# Almacén de sesiones en el SERVIDOR: session_id -> datos
SESIONES: dict[str, dict] = {}


class Credenciales(BaseModel):
    usuario: str = Field(..., min_length=1, max_length=50)
    password: str = Field(..., min_length=1, max_length=128)


def sesion_actual(session_id: str | None) -> dict:
    datos = SESIONES.get(session_id or "")
    if not datos or datos["expira"] < time.time():
        SESIONES.pop(session_id or "", None)      # limpia sesiones vencidas
        raise HTTPException(status.HTTP_401_UNAUTHORIZED, "Sesión inválida o expirada")
    return datos


@app.post("/api/login")
def login(cred: Credenciales, response: Response) -> dict:
    usuario = USUARIOS.get(cred.usuario)
    # compare_digest compara en tiempo constante
    if not usuario or not hmac.compare_digest(usuario["password"], cred.password):
        raise HTTPException(status.HTTP_401_UNAUTHORIZED, "Usuario o contraseña incorrectos")

    session_id = secrets.token_urlsafe(32)         # id aleatorio e impredecible
    SESIONES[session_id] = {"usuario": cred.usuario, "expira": time.time() + DURACION_SEG}
    response.set_cookie(
        key=COOKIE, value=session_id,
        httponly=True,           # JavaScript no puede leerla (mitiga robo por XSS)
        samesite="lax",          # no viaja en peticiones POST de otros sitios (mitiga CSRF)
        max_age=DURACION_SEG,
        secure=False,            # en producción con HTTPS: True (aquí el TestClient usa http)
        path="/",
    )
    return {"usuario": cred.usuario, "mensaje": "Sesión iniciada"}


@app.get("/api/perfil")
def perfil(session_id: str | None = Cookie(default=None)) -> dict:
    datos = sesion_actual(session_id)
    return {"usuario": datos["usuario"], "nombre": USUARIOS[datos["usuario"]]["nombre"]}


@app.post("/api/logout", status_code=status.HTTP_204_NO_CONTENT)
def logout(response: Response, session_id: str | None = Cookie(default=None)) -> Response:
    SESIONES.pop(session_id or "", None)           # invalida en el servidor
    respuesta = Response(status_code=status.HTTP_204_NO_CONTENT)
    respuesta.delete_cookie(COOKIE, path="/", httponly=True, samesite="lax")
    return respuesta
````
