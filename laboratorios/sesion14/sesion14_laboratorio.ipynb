### 📄 `app.py`

Archivo principal de la implementación.

````python
%%writefile app.py
"""Sesión 14 · Registro y login con bcrypt: solo se guarda el hash, nunca la contraseña."""
import bcrypt
from fastapi import FastAPI, HTTPException, status
from pydantic import BaseModel, Field, field_validator

app = FastAPI(title="Hashing seguro · DAB-2201", version="1.0.0")

COSTO = 12                                   # factor de trabajo: 2^12 rondas (más alto = más lento)
_usuarios: dict[str, bytes] = {}             # usuario -> hash bcrypt (incluye el salt)
_HASH_SEÑUELO = bcrypt.hashpw(b"no-existe", bcrypt.gensalt(COSTO))


class Credenciales(BaseModel):
    usuario: str = Field(..., min_length=3, max_length=30, pattern=r"^[a-zA-Z0-9_.-]+$")
    password: str = Field(..., min_length=6)

    @field_validator("password")
    @classmethod
    def maximo_72_bytes(cls, v: str) -> str:
        # bcrypt solo usa los primeros 72 BYTES; mejor rechazar que truncar en silencio
        if len(v.encode("utf-8")) > 72:
            raise ValueError("la contraseña no puede exceder 72 bytes")
        return v


class UsuarioPublico(BaseModel):
    """Lo único que sale hacia el cliente: nunca la contraseña ni su hash."""
    usuario: str
    mensaje: str


@app.post("/api/registro", status_code=status.HTTP_201_CREATED, response_model=UsuarioPublico)
def registro(cred: Credenciales) -> UsuarioPublico:
    if cred.usuario in _usuarios:
        # 409: el usuario existe; NO se sobrescribe lo guardado
        raise HTTPException(status.HTTP_409_CONFLICT, "El usuario ya está registrado")
    _usuarios[cred.usuario] = bcrypt.hashpw(cred.password.encode("utf-8"), bcrypt.gensalt(COSTO))
    return UsuarioPublico(usuario=cred.usuario, mensaje="Cuenta creada")


@app.post("/api/login")
def login(cred: Credenciales) -> dict:
    guardado = _usuarios.get(cred.usuario)
    # Si el usuario no existe se verifica contra un señuelo para tardar lo mismo
    # (así no se puede adivinar qué usuarios existen midiendo el tiempo de respuesta).
    valido = bcrypt.checkpw(cred.password.encode("utf-8"), guardado or _HASH_SEÑUELO)
    if not guardado or not valido:
        raise HTTPException(status.HTTP_401_UNAUTHORIZED, "Usuario o contraseña incorrectos")
    return {"usuario": cred.usuario, "autenticado": True}
````
